```python
# ============================================
# IDENTIFY TARGET COLUMN
# ============================================

print("\nAvailable columns:")
for i, column in enumerate(data.columns):
    print(i, ":", column)

target = data.columns[-1]

print("\nTarget column selected:", target)
```

```python
# ============================================
# IDENTIFY TARGET COLUMN
# ============================================

print("\nAvailable columns:")
for i, column in enumerate(data.columns):
    print(i, ":", column)

target = data.columns[-1]

print("\nTarget column selected:", target)
```

```python
# ============================================
# DATA PREPROCESSING
# ============================================

# Remove Day if it exists
if "Day" in data.columns:
    data = data.drop("Day", axis=1)

# Remove rows containing missing values
data = data.dropna()

print("\nDataset after preprocessing:")
print(data)

print("\nNumber of rows:", len(data))
```

```python
# ============================================
# FIND-S ALGORITHM
# ============================================

# Get attribute columns
attributes = list(data.columns)

# Remove target column
attributes.remove(target)

print("\nAttributes:")
print(attributes)

# Start with most specific hypothesis
hypothesis = ["∅"] * len(attributes)

print("\nInitial Hypothesis:")
print(hypothesis)


# ============================================
# PROCESS TRAINING EXAMPLES
# ============================================

for index, row in data.iterrows():

    # Process only positive examples
    if str(row[target]).strip().lower() == "yes":

        print("\n----------------------------------------")
        print("Positive Example:", index + 1)
        print("----------------------------------------")

        for i in range(len(attributes)):

            # First positive example
            if hypothesis[i] == "∅":
                hypothesis[i] = row[attributes[i]]

            # If values are different,
            # generalize the hypothesis
            elif hypothesis[i] != row[attributes[i]]:
                hypothesis[i] = "?"

        print("Current Hypothesis:")
        print(hypothesis)
```

```python
# ============================================
# FINAL HYPOTHESIS
# ============================================

print("\n============================================")
print("FINAL HYPOTHESIS")
print("============================================")

for attribute, value in zip(attributes, hypothesis):
    print(attribute, "=", value)

print("\nMost Specific Hypothesis:")
print(hypothesis)
```

```python
import pandas as pd
import random

random.seed(42)

outlook = ["Sunny", "Overcast", "Rain"]
temp = ["Hot", "Mild", "Cool"]
humidity = ["High", "Normal"]
wind = ["Weak", "Strong"]

data = []

for i in range(100):
    o = random.choice(outlook)
    t = random.choice(temp)
    h = random.choice(humidity)
    w = random.choice(wind)

    # Simple realistic rule for Play Tennis
    if o == "Overcast":
        play = "Yes"
    elif o == "Sunny" and h == "High":
        play = "No"
    elif o == "Rain" and w == "Strong":
        play = "No"
    else:
        play = "Yes"

    data.append([o, t, h, w, play])

df = pd.DataFrame(
    data,
    columns=["outlook", "temp", "humidity", "wind", "play"]
)

print("Dataset created successfully!")
print("Shape:", df.shape)

df.head(10)
```

```python
print("Number of rows:", len(df))
print("\nColumns:")
print(df.columns.tolist())

print("\nPlay distribution:")
print(df["play"].value_counts())

print("\nComplete dataset:")
display(df)
```
